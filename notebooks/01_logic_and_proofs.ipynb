# Logic and Proofs

This notebook covers propositional logic, logical connectives, truth tables, and proof techniques in discrete mathematics.

## 1. Propositional Logic

A **proposition** is a declarative statement that is either true or false. We combine propositions using logical connectives: NOT (¬), AND (∧), OR (∨), IMPLIES (→), and BICONDITIONAL (↔).

```python
from itertools import product

def truth_table(variables, expr):
    """Print a truth table for a boolean expression.

    variables: list of variable names
    expr: function taking a dict of variable->bool and returning bool
    """
    header = variables + ["result"]
    print(" | ".join(f"{h:>6}" for h in header))
    print("-" * (9 * len(header)))
    for values in product([False, True], repeat=len(variables)):
        env = dict(zip(variables, values))
        result = expr(env)
        row = [str(env[v]) for v in variables] + [str(result)]
        print(" | ".join(f"{c:>6}" for c in row))

# Example: implication p -> q
truth_table(["p", "q"], lambda e: (not e["p"]) or e["q"])
```

## 2. Logical Equivalences

Two propositions are **logically equivalent** if they have the same truth value in every case. We verify De Morgan's law: ¬(p ∧ q) ≡ ¬p ∨ ¬q.

```python
def equivalent(variables, expr1, expr2):
    for values in product([False, True], repeat=len(variables)):
        env = dict(zip(variables, values))
        if expr1(env) != expr2(env):
            return False
    return True

de_morgan_lhs = lambda e: not (e["p"] and e["q"])
de_morgan_rhs = lambda e: (not e["p"]) or (not e["q"])

print("De Morgan's law holds:", equivalent(["p", "q"], de_morgan_lhs, de_morgan_rhs))
```

## 3. Tautologies and Contradictions

A **tautology** is always true; a **contradiction** is always false.

```python
def classify(variables, expr):
    results = [expr(dict(zip(variables, v)))
               for v in product([False, True], repeat=len(variables))]
    if all(results):
        return "tautology"
    if not any(results):
        return "contradiction"
    return "contingency"

print("p ∨ ¬p :", classify(["p"], lambda e: e["p"] or not e["p"]))
print("p ∧ ¬p :", classify(["p"], lambda e: e["p"] and not e["p"]))
```

## 4. Proof Techniques

Common techniques: **direct proof**, **proof by contraposition**, and **proof by contradiction**. Below we demonstrate by checking a claim computationally before a formal argument.

**Claim:** If $n$ is even then $n^2$ is even.

**Direct proof:** Let $n = 2k$. Then $n^2 = 4k^2 = 2(2k^2)$, which is even. ∎

```python
def n_squared_even_check(limit=1000):
    for n in range(limit):
        if n % 2 == 0 and (n * n) % 2 != 0:
            return False
    return True

print("Empirical check passes:", n_squared_even_check())
```
