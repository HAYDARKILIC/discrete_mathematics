# Set Theory

Sets, set operations, Venn diagram concepts, power sets, and Cartesian products.

## 1. Basic Set Operations

Union, intersection, difference, and symmetric difference.

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("A ∪ B :", A | B)
print("A ∩ B :", A & B)
print("A - B :", A - B)
print("A △ B :", A ^ B)
```

## 2. Subsets and Power Sets

The **power set** of a set $S$ is the set of all subsets of $S$. If $|S| = n$, then $|P(S)| = 2^n$.

```python
from itertools import chain, combinations

def power_set(s):
    s = list(s)
    return list(chain.from_iterable(combinations(s, r) for r in range(len(s) + 1)))

S = {1, 2, 3}
ps = power_set(S)
print(f"Power set of {S}:")
for subset in ps:
    print(set(subset) if subset else "∅")
print(f"\nSize: {len(ps)} = 2^{len(S)}")
```

## 3. Cartesian Product

The Cartesian product $A \\times B$ is the set of all ordered pairs $(a, b)$ with $a \\in A$ and $b \\in B$.

```python
from itertools import product as cartesian

A = {1, 2}
B = {'x', 'y', 'z'}
prod = list(cartesian(A, B))
print("A × B :", prod)
print("|A × B| =", len(prod), "= |A| · |B| =", len(A) * len(B))
```

## 4. Set Identities

Verify De Morgan's laws for sets within a universal set $U$.

```python
U = set(range(1, 11))
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

lhs = U - (A | B)
rhs = (U - A) & (U - B)
print("(A ∪ B)' =", lhs)
print("A' ∩ B'  =", rhs)
print("De Morgan holds:", lhs == rhs)
```
