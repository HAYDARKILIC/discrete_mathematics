# Relations and Functions

Binary relations, properties (reflexive, symmetric, transitive), equivalence relations, and functions.

## 1. Binary Relations

A binary relation $R$ on a set $A$ is a subset of $A \\times A$. We represent it as a set of pairs.

```python
A = {1, 2, 3}
R = {(1, 1), (2, 2), (3, 3), (1, 2), (2, 1)}

def is_reflexive(A, R):
    return all((a, a) in R for a in A)

def is_symmetric(R):
    return all((b, a) in R for (a, b) in R)

def is_transitive(R):
    return all((a, c) in R for (a, b) in R for (c, d) in R if b == c)

print("Reflexive: ", is_reflexive(A, R))
print("Symmetric: ", is_symmetric(R))
print("Transitive:", is_transitive(R))
```

## 2. Equivalence Relations

An **equivalence relation** is reflexive, symmetric, and transitive. It partitions the set into equivalence classes.

```python
def is_equivalence(A, R):
    return is_reflexive(A, R) and is_symmetric(R) and is_transitive(R)

def equivalence_classes(A, R):
    classes = []
    seen = set()
    for a in A:
        if a in seen:
            continue
        cls = {b for b in A if (a, b) in R}
        classes.append(cls)
        seen |= cls
    return classes

print("Is equivalence relation:", is_equivalence(A, R))
print("Equivalence classes:", equivalence_classes(A, R))
```

## 3. Functions

A **function** $f: A \\to B$ assigns exactly one element of $B$ to each element of $A$. We study injective, surjective, and bijective functions.

```python
def is_injective(f, domain):
    images = [f(x) for x in domain]
    return len(images) == len(set(images))

def is_surjective(f, domain, codomain):
    return set(f(x) for x in domain) == set(codomain)

domain = [1, 2, 3]
codomain = [2, 4, 6]
f = lambda x: 2 * x

print("Injective: ", is_injective(f, domain))
print("Surjective:", is_surjective(f, domain, codomain))
print("Bijective: ", is_injective(f, domain) and is_surjective(f, domain, codomain))
```
