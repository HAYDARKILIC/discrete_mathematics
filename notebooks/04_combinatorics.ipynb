# Combinatorics

Counting principles, permutations, combinations, the binomial theorem, and the pigeonhole principle.

## 1. Permutations and Combinations

$P(n, r) = \\frac{n!}{(n-r)!}$ counts ordered arrangements; $C(n, r) = \\frac{n!}{r!(n-r)!}$ counts unordered selections.

```python
from math import factorial, comb, perm

def permutations_count(n, r):
    return factorial(n) // factorial(n - r)

def combinations_count(n, r):
    return factorial(n) // (factorial(r) * factorial(n - r))

print("P(5, 2) =", permutations_count(5, 2), "| built-in:", perm(5, 2))
print("C(5, 2) =", combinations_count(5, 2), "| built-in:", comb(5, 2))
```

## 2. The Binomial Theorem

$(x + y)^n = \\sum_{k=0}^{n} \\binom{n}{k} x^{n-k} y^k$

```python
def binomial_expansion(n):
    terms = []
    for k in range(n + 1):
        coeff = comb(n, k)
        terms.append(f"{coeff}·x^{n-k}·y^{k}")
    return " + ".join(terms)

print(f"(x + y)^4 = {binomial_expansion(4)}")
```

## 3. Pascal's Triangle

Each entry is the sum of the two entries above it, corresponding to binomial coefficients.

```python
def pascals_triangle(rows):
    triangle = []
    for n in range(rows):
        triangle.append([comb(n, k) for k in range(n + 1)])
    return triangle

for row in pascals_triangle(6):
    print(" ".join(f"{x:>3}" for x in row).center(40))
```

## 4. Pigeonhole Principle

If $n$ items are placed into $m$ containers with $n > m$, then at least one container holds more than one item.

```python
def pigeonhole_min_collisions(items, containers):
    # Minimum number guaranteed in the fullest container
    return -(-items // containers)  # ceiling division

print("13 people, 12 months -> at least",
      pigeonhole_min_collisions(13, 12), "share a birth month")
```
