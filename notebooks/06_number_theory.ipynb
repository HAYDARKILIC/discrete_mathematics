# Number Theory

Divisibility, the Euclidean algorithm, modular arithmetic, and primes.

## 1. Greatest Common Divisor

The **Euclidean algorithm** computes $\\gcd(a, b)$ efficiently.

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a

def lcm(a, b):
    return a * b // gcd(a, b)

print("gcd(48, 36) =", gcd(48, 36))
print("lcm(4, 6)   =", lcm(4, 6))
```

## 2. Extended Euclidean Algorithm

Finds integers $x, y$ such that $ax + by = \\gcd(a, b)$ (Bézout's identity).

```python
def extended_gcd(a, b):
    if b == 0:
        return a, 1, 0
    g, x1, y1 = extended_gcd(b, a % b)
    return g, y1, x1 - (a // b) * y1

g, x, y = extended_gcd(48, 36)
print(f"gcd = {g}, x = {x}, y = {y}")
print(f"Check: 48·{x} + 36·{y} = {48*x + 36*y}")
```

## 3. Modular Arithmetic

Modular exponentiation and modular inverses are central to cryptography.

```python
def mod_inverse(a, m):
    g, x, _ = extended_gcd(a, m)
    if g != 1:
        return None  # inverse does not exist
    return x % m

print("3^4 mod 5 =", pow(3, 4, 5))
print("Inverse of 3 mod 7 =", mod_inverse(3, 7))
```

## 4. Primality and the Sieve of Eratosthenes

Generate all primes up to $n$ efficiently.

```python
def sieve(n):
    is_prime = [True] * (n + 1)
    is_prime[0] = is_prime[1] = False
    for i in range(2, int(n ** 0.5) + 1):
        if is_prime[i]:
            for j in range(i * i, n + 1, i):
                is_prime[j] = False
    return [i for i, p in enumerate(is_prime) if p]

print("Primes up to 50:", sieve(50))
```
