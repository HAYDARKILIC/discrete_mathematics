# Boolean Algebra

Boolean functions, simplification, normal forms, and logic gates.

## 1. Boolean Functions and Gates

Basic gates: AND, OR, NOT, NAND, NOR, XOR.

```python
def AND(a, b): return a and b
def OR(a, b):  return a or b
def NOT(a):    return not a
def NAND(a, b): return not (a and b)
def XOR(a, b):  return a != b

from itertools import product
print("a  b | AND OR XOR NAND")
for a, b in product([0, 1], repeat=2):
    print(f"{a}  {b} |  {int(AND(a,b))}   {int(OR(a,b))}   {int(XOR(a,b))}    {int(NAND(a,b))}")
```

## 2. Disjunctive Normal Form (DNF)

Any Boolean function can be expressed as an OR of ANDs (sum of products) derived from its truth table.

```python
def to_dnf(variables, func):
    terms = []
    for values in product([0, 1], repeat=len(variables)):
        if func(*values):
            literals = []
            for var, val in zip(variables, values):
                literals.append(var if val else f"¬{var}")
            terms.append("(" + " ∧ ".join(literals) + ")")
    return " ∨ ".join(terms) if terms else "0"

# Example: XOR of two variables
print("XOR DNF:", to_dnf(["a", "b"], lambda a, b: a != b))
```

## 3. Functional Completeness

NAND alone can express all Boolean functions. We build NOT, AND, and OR from NAND.

```python
def not_from_nand(a):       return NAND(a, a)
def and_from_nand(a, b):    return NAND(NAND(a, b), NAND(a, b))
def or_from_nand(a, b):     return NAND(NAND(a, a), NAND(b, b))

for a, b in product([0, 1], repeat=2):
    assert and_from_nand(a, b) == AND(a, b)
    assert or_from_nand(a, b) == OR(a, b)
print("NAND is functionally complete: verified ✓")
```
